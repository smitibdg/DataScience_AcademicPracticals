```python
import nltk
from nltk.corpus import wordnet
```

```python
nltk.download('wordnet')
```

```python
nltk.download('omw-1.4')
```

```python
word = "active"
synonyms = set()
antonynms = set()
```

```python
for synset in wordnet.synsets(word):
  for lemma in synset.lemmas():
    synonyms.add(lemma.name())
    if lemma.antonyms():
      antonynms.add(lemma.antonyms()[0].name())

print("Word: ", word)
```

```python
print("synonyms", synonyms)
for s in sorted(synonyms):
  print(s)

print("antonynms", antonynms)
for a in sorted(antonynms):
  print(a)
```
