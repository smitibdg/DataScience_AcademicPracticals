```python
import pandas as pd
```

```python
def smiti_minimum_edit_distance(s1, s2, insert_cost_smiti=1, delete_cost_smiti=1, substitute_cost=1):
  m = len(s1)
  n = len(s2)

  dp = [[0 for _ in range(n+1)] for _ in range(m+1)]

  for i in range(m+1):
    dp[i][0] = i * delete_cost_smiti
  for j in range(n+1):
    dp[0][j] = j * insert_cost_smiti

  for i in range(1, m+1):
    for j in range(1, n+1):
      if s1[i-1] == s2[j-1]:
        cost = 0
      else:
        cost = substitute_cost

      dp[i][j] = min(
          dp[i-1][j] + delete_cost_smiti,
          dp[i][j-1] + insert_cost_smiti,
          dp[i-1][j-1] + cost
      )
  return dp, dp[m][n]

def show_dp_table(s1, s2, dp):
  df = pd.DataFrame(dp, index=[''] + list(s1), columns=[''] + list(s2))
  display(df)
```

**Test Cases**

```python
test_cases = [
    ("kitten", "sitting"),        #Substitution + insertion
    ("book", "back"),             #Substitution
    ("intention", "execution"),   #Complex changes
    ("speling", "spelling"),      #typo(insertion)
    ("playing", "play"),          # deletion
]

for s1, s2 in test_cases:
  dp_table, distance = smiti_minimum_edit_distance(s1, s2)
  print(f"Source String: {s1}")
  print(f"Target String: {s2}")
  print(f"Minimum Edit Distance: {distance}")
  show_dp_table(s1, s2, dp_table)
```

**Adaptability Test**

```python
print("Adaptability Test (Different Costs)")
s1 = "colour"
s2 = "color"

dp_table, distance = smiti_minimum_edit_distance(s1, s2, substitute_cost=2)
print("Source: ", s1)
print("Target: ", s2)
print("Edit Distance(Substitution Cost = 2): ", distance)
show_dp_table(s1, s2, dp_table)
```
