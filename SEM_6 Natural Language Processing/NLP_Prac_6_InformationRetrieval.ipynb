2. Study Various Corpora

Brown, Inaugural. Reuters, UDHR

```python
from nltk.corpus import brown, inaugural, reuters, udhr
# Brown Corpus
print("Brown Categories_smiti: ", brown.categories()[:5])
print("Brown Words_smiti: ", brown.words()[:10])
print("Brown Sentences_smiti: ", brown.sents()[0])

# Inaugural Corpus
print("\nInaugural Fileids_smiti: ", inaugural.fileids()[:5])
print("Inaugural Words_smiti: ", inaugural.words()[:10])

# Reuters Corpus
print("\nReuters Categories_smiti: ", reuters.fileids()[:5])
print("Reuters Words_smiti: ", reuters.words()[:10])

# UDHR Corpus
print("\nUDHR Fileids_smiti: ", udhr.fileids()[:5])
print("UDHR Words_smiti: ", udhr.words()[:10])
```

4. Conditional Frequency Distribution(CFD)

```python
from nltk import ConditionalFreqDist

cfd_smiti = ConditionalFreqDist(
    (genre, word)
    for genre in brown.categories()
    for word in brown.words(categories=genre)
)

print("Frequency of 'the' in news: ", cfd_smiti['news']['the'])
print("Frequency of 'the' in romance: ", cfd_smiti['romance']['the'])
```

5. Study the tagged Corpora

Tagged Sentences and Tagged words

```python
# Tagged Sentences
tagged_sent_smiti = brown.tagged_sents()[0]
print("Tagged Sentences: ", tagged_sent_smiti)

# Tagged Words
tagged_word_smiti = brown.tagged_words()[0]
print("Tagged Words: ", tagged_word_smiti)
```

6. Program to find most frequent noun tags

```python
from collections import Counter

nouns_smiti = [tag for (word, tag) in brown.tagged_words() if tag.startswith('NN')]
noun_freq_smiti = Counter(nouns_smiti)

print("Most Common Noun Tags: ", noun_freq_smiti.most_common(5))
```

8. Rule-Based Tagger and Unigram

```python
from nltk.tag import RegexpTagger, UnigramTagger

# Rule-based Tagger
patterns_smiti = [
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*es$', 'VBZ'),
    (r'.*', 'NN')
]

rule_tagger_smiti = RegexpTagger(patterns_smiti)
print(rule_tagger_smiti.tag("He is running fast".split()))

# Unigram tagger
train_sents_smiti = brown.tagged_sents(categories="news")
unigram_tagger_smiti = UnigramTagger(train_sents_smiti)

print(unigram_tagger_smiti.tag("The quick brown fox jumps".split()))
```

9. Find Different Words from Plain Text & Score Words

```python
from nltk.corpus import stopwords
from collections import Counter

words_smiti = brown.words()
filtered_words_smiti = [w.lower() for w in words_smiti if w.isalpha()]

freq_dist = Counter(filtered_words_smiti)

print("Most Common Words: ", freq_dist.most_common(5))
```
