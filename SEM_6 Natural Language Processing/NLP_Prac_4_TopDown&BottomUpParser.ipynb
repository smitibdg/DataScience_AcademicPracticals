**Top-Down Parser**

```python
class TopDownParser:
  def __init__(self, input_string):
    self.input = input_string
    self.index = 0

  def match(self, char):
    if self.index < len(self.input) and self.input[self.index] == char:
      self.index += 1
      return True
    return False

  def S(self):
    start = self.index

    # Rule: S - a S b
    if self.match('a'):
      if self.S() and self.match('b'):
          return True
      self.index = start # backtrack

    # Rule: S - E (e: epsilon)
    return True

  def parse(self):
    result = self.S()
    return result and self.index == len(self.input)

# Testing
test_strings = ["", "ab", "aabb", "aaabbb", "aab", "abb"]

print("Top-Down Parser Results:")
for s in test_strings:
    parser = TopDownParser(s)
    print(f"Input: '{s}' - Accepted: {parser.parse()}")
```

**Bottom Up Parser**

```python
def bottom_up_parser(input_string):
  stack = []
  i = 0

  while True:
    # SHIFT if input remains
    if i < len(input_string):
      stack.append(input_string[i])
      i += 1

    # REDUCE repeatedly
    reduced = True
    while reduced:
      reduced = False
      stack_str = ''.join(stack)

      # Reduce aSb - S
      if 'aSb' in stack_str:
        stack_str = stack_str.replace('aSb', 'S', 1)
        stack = list(stack_str)
        reduced = True

      # Reduce ab - S
      elif 'ab' in stack_str:
        stack_str = stack_str.replace('ab', 'S', 1)
        stack = list(stack_str)
        reduced = True

    # Stop condition
    if i >= len(input_string):
      break
  return ''.join(stack) == 'S' or input_string == ""

# Testing
print("\nBottom-Up Parser Results:")
for s in test_strings:
  print(f"Input: '{s}' - Accepted: {bottom_up_parser(s)}")
```
