```python
import re
txt = "Hello World"
```

```python
# Findall will take pattern and text
# Any 1 must be present in the given range
x = re.findall("[a-m]", txt) #Find all lowercase characters alphabetically from 'a' to 'b'.
print(x)
```

```python
# Start from "H"
# .. can be character except \n(new line)
# end with 'o'
x = re.findall('He..o', txt)# Search for a sequence that starts with "he", followed by two (any) characters, and an 'o'
print(x)
```

```python
x = re.findall("^Hello", txt) # Check if the string starts with 'Hello'
if x:
  print("Yes, the string starts with 'Hello'")
else:
  print("No match")
```

```python
x = re.findall("World$", txt) # Check if the string ends with 'planet':
if x:
  print("Yes, the string ends with 'World'")
else:
  print("No match")
```

```python
# Search for a sequence that starts with 'he', followed by 0 or more (any) characters, and an 'r'
x = re.findall("He.*r", txt)
print(x)
```

```python
# Search for a sequence that starts with 'he', followed by 0 or more (any) characters, and ends with 'o'
x = re.findall("He.+o", txt)
print(x)
```

```python
# Search for a sequence that starts with 'he', followed by 0 or 1 (any) character, and an 'o'
x = re.findall("He.?o", txt)
print(x) # This time we will get no match, because there were not one, but two characters between 'he' and the 'o'.
```

```python
# Search for a sequence that
x = re.findall("He.{2}o", txt)
print(x)
```

```python
txt = "The rain in Spain falls mainly in the plain!"
# Check if the string contains either "falls or "stays"
x = re.findall("falls|stays", txt)
print(x)
if x:
  print("Yes, there is at least one match!")
else:
  print("No match")
```

```python
import re
txt = "Åland"
# Find all ASCII matches
print(re.findall("\w", txt, re.ASCII))

# Without the flag, the example would return all character
print(re.findall("\w", txt))

# Same result using the shorthand re.A flag
print(re.findall("\w", txt, re.A))
```

```python
txt = "The rain in Spain"
# Use a case insensitive search when finding a match for Spain in the text
print(re.findall("spain", txt, re.DEBUG))
```

```python
txt = """hi
my
name
is
Rizz"""

# Search for a sequence that starts with 'me', followed by one character, even a new line character and continuos with "is"
print(re.findall("me.is", txt, re.DOTALL))

# This example would return no match the re.DOTALL flag
print(re.findall("me.is", txt))

# Same result with the shorthand re.S flag:
print(re.findall("me.is", txt, re.S))
```

```python
txt = "The rain in Spain"
# Use a case insensitive search when finding a match for Spain in the text
print(re.findall("spain", txt, re.IGNORECASE))

# Same result using the shorthand re.I flag:
print(re.findall("spain", txt, re.I))
```

```python
txt = """There
aint much
rain in
Spain"""

# Search for a sequence "ain", at the begining of the line:
print(re.findall("^ain", txt, re.MULTILINE))

# This example would return no matches without the re.MULTILINE flag
print(re.findall("^ain", txt))

# Same result with the shorthand re.M flag:
print(re.findall("^ain", txt, re.M))
```

```python
text = "The rain in Spain falls mainly on the plain"

#Find and return words that contains the phrase "ain"
pattern = """
[A-Za-z]* #Starts with any letter
ain+      #contains 'ain'
[a-z]*    #followed by any small letter
"""

print(re.findall(pattern, text, re.VERBOSE))

#The example would return nothing without the re.VERBOSE flag:
print(re.findall(pattern, text))

# Same result with the shorthand re.X flag:
print(re.findall(pattern, text, re.X))
```
