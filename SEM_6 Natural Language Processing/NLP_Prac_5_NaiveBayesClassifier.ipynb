```python
# Naive Bayes Test Classification with Add-1 (Laplace) Smoothing

from collections import Counter
import math

# Training data

documents = [
    ("fun couple love love", "comedy"),
    ("fast furious shoot", "action"),
    ("couple fly fast fun fun", "comedy"),
    ("furious shoot shoot fun", "action"),
    ("fly fast shoot love", "action")
]
```

```python
# New document
test_doc = "fast couple shoot fly"
```

```python
# Preprocessing
def tokenize(text):
  return text.lower().split()
```

```python
# Separate documents by class
class_docs = {"comedy": [], "action": []}
```

```python
for text, label in documents:
  class_docs[label].extend(tokenize(text))
```

```python
# Vocabulary
vocab = set()
for words in class_docs.values():
  vocab.update(words)
V = len(vocab)
```

```python
# Prior probabilities
total_docs = len(documents)
class_counts = Counter(label for _, label in documents)
priors = {
    cls: class_counts[cls] / total_docs
    for cls in class_counts
}
```

```python
# Likelihood with add-1 smoothing
word_counts = {
    cls: Counter(class_docs[cls])
    for cls in class_docs
    }
total_words = {
    cls: sum(word_counts[cls].values())
    for cls in word_counts
}
```

```python
def likelihood(word,cls):
  return (word_counts[cls][word] + 1) / (total_words[cls] + V)
```

```python
# Classification
test_words = tokenize(test_doc)
scores = {}
```

```python
for cls in priors:
  log_prob = math.log(priors[cls])
  for word in test_words:
      log_prob += math.log(likelihood(word,cls))
  scores[cls] = log_prob
```

```python
# Output
print("Vocabulary: ", vocab)
print("Prior probabilities: ", priors)
for cls, p in scores.items():
  print(f"P({cls}) = {p}")
print("Log Probabilities for Test Document:")
for cls, score in scores.items():
  print(f"P({cls}: {score}")
predicted_class = max(scores, key=scores.get)
print("Predicted Class for document D:", predicted_class)
```
