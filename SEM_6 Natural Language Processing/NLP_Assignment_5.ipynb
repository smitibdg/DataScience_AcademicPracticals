Q.4. Write expressions for finding all words in text6 that meet the conditions listed below.

The result should be in the form of a list of words: ['word1', 'word2', ...].

*   Ending in ise
*   Containing the letter z
*   Containing the sequence of letters pt
*   Having all lowercase letters except for an initial capital (i.e., titlecase)

```python
[w for w in text6 if w[-3:] == 'ise']
```

```python
[w for w in text6 if 'i' in w]
```

```python
[w for w in text6 if 'pt' in w]
```

```python
[w for w in text6 if w.istitle()]
```

Q.5. Define sent to be the list of words ['she', 'sells', 'sea', 'shells', 'by', 'the', 'sea', 'shore'].

Now write code to perform the following tasks:

*   Print all words begining with ah
*   Print all words longer than four characters

```python
sent = ['she', 'sells', 'sea', 'shells', 'by', 'the', 'sea', 'shore']
```

```python
[w for w in sent if w[0:2] == 'ah']
```

```python
[w for w in sent if len(w) > 4]
```

Q.6. What does the following Python code do? sum(len(w) for w in text1) Can you use it work out the average word length of a text?

```python
(sum(len(w) for w in text1))/len(text1)
```

The given code sums up all the letters of the words in text i.e. it gives us the vocabulary size of text Yes, it can be used to find the average word length as done above.

Q.7 Define a function called vocab_size(text) that has a single parameter for the text, and which returns the vocabulary size of the text.

```python
def vocab_size(t):
  return sum(len(w) for w in t)
```
