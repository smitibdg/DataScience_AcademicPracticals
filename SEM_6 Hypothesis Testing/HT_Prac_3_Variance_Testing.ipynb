# a) Chi-square Test of a Single Variance
## Purpose

To test whether the population variance is equal to a given (hypothesized) variance.

## Hypotheses

H₀: σ² = σ₀²

H₁: σ² ≠ σ₀²

```python
import numpy as np
import scipy.stats as stats
```

```python
alpha = 0.05
# Sample data
data_smiti = np.array([12, 15, 14, 10, 13, 16, 14, 11, 15, 14])
n_smiti = len(data_smiti)
```

```python
sample_variance = np.var(data_smiti, ddof=1)
hypothesized_variance = 4
```

```python
chi_square_stat = (n_smiti - 1) * sample_variance / hypothesized_variance
chi_lower = stats.chi2.ppf(alpha/2, n_smiti-1)
chi_upper = stats.chi2.ppf(1 - alpha/2, n_smiti-1)
```

```python
print("Chi-square Test for Single Variance")
print("Sample Variance:", sample_variance)
print("Chi-square Statistic:", chi_square_stat)
print("Critical Values:", chi_lower, "and", chi_upper)
print("Reject H0:", chi_square_stat < chi_lower or chi_square_stat > chi_upper)
```

# b) F-test for Two Variances
## Purpose

To compare variances of two populations.

## Hypotheses

H₀: σ₁² = σ₂²

H₁: σ₁² ≠ σ₂²

```python
# Sample data
sample1_smiti = np.array([10, 12, 14, 11, 13])
sample2_smiti = np.array([8, 9, 7, 10, 9])
```

```python
var1 = np.var(sample1_smiti, ddof=1)
var2 = np.var(sample2_smiti, ddof=1)
```

```python
F_stat = var1 / var2
```

```python
df1 = len(sample1_smiti) - 1
df2 = len(sample2_smiti) - 1
```

```python
F_critical = stats.f.ppf(1 - alpha, df1, df2)
```

```python
print("\nF-test for Two Variances")
print("Variance 1:", var1)
print("Variance 2:", var2)
print("F Statistic:", F_stat)
print("F Critical Value:", F_critical)
print("Reject H0:", F_stat > F_critical)
```

# c) Testing of Homogeneity of Variances
## Purpose

To check whether multiple groups have equal variances.

## Test Used

✔ Levene’s Test

## Hypotheses

H₀: All group variances are equal

H₁: At least one variance is different

```python
# Data from three groups
group1_smiti = np.array([20, 22, 21, 23, 24])
group2_smiti = np.array([30, 32, 31, 29, 28])
group3_smiti = np.array([40, 42, 41, 39, 38])
```

```python
levene_stat, p_value = stats.levene(group1_smiti, group2_smiti, group3_smiti)
```

```python
print("\nTest of Homogeneity of Variances (Levene’s Test)")
print("Levene Statistic:", levene_stat)
print("P-value:", p_value)
print("Reject H0:", p_value < alpha)
```

# Interpretation

## Chi-square test checks if a sample variance matches a hypothesized variance.

## F-test compares two population variances.

## Levene’s test checks equality of variances across multiple groups.

If the test statistic lies in the rejection region (or p-value < α), the null hypothesis is rejected.

Variance testing is used to test equality of population variances using Chi-square test, F-test, and homogeneity tests such as Levene’s test.
