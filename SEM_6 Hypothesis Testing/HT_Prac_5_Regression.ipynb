SIMPLE LINEAR REGRESSION

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Samples Data
np.random.seed(1)
x = np.random.randint(10, 100, 50).reshape(-1, 1) # Advertising budget
y = 3 * x.flatten() + 10 + np.random.normal(0, 20, 50) #noise(0 is mean and 20 is S.D.)
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
y_pred = model.predict(x)
```

```python
print("Slope (b1): ", model.coef_[0])
print("Intercept (b0): ", model.intercept_)
print("R2 Score: ",r2_score(y, y_pred))
print("RMSE: ", np.sqrt(mean_squared_error(y, y_pred)))
```

```python
plt.scatter(x, y)
plt.plot(x, y_pred)
plt.xlabel("Advertising Budget")
plt.ylabel("Sales")
plt.title("Simple Linear Regression")
plt.show()
```

MULTIPLE LINEAR REGRESSION

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
np.random.seed(2)
# Sample Data
size = np.random.randint(600, 3000, 50)   # House size
bedrooms = np.random.randint(1, 5, 50)    # Number of bedrooms
age = np.random.randint(1, 30, 50)        # Age of house
```

```python
# House price formula + noise
price = 50*size + 20000*bedrooms + 10000*age + np.random.normal(0, 50000, 50)
```

```python
# Convert to DataFrame
df = pd.DataFrame({
    'size': size,
    'bedrooms': bedrooms,
    'age': age,
    'price': price
})
```

```python
x = df[['size', 'bedrooms', 'age']] # Multiple Independent
y = df[['price']]
```

```python
# Build model
model = LinearRegression()
model.fit(x, y)
```

```python
# Prediction
y_pred = model.predict(x)
```

```python
# Output
print("Coefficiants: ", model.coef_)
print("Intercept: ", model.intercept_)
print("R2 Score: ", r2_score(y, y_pred))
print("RMSE: ", np.sqrt(mean_squared_error(y, y_pred)))
```

POLYNOMIAL REGRESSION

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Samples Data
np.random.seed(3)
x = np.linspace(0, 10, 50).reshape(-1, 1)
y = 2 + 1.5*x.flatten() ** 2 + 5 * np.random.randn(50)
```

```python
# Create polynomial Features (degree=2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(x)
```

```python
# Build Model
model = LinearRegression()
model.fit(X_poly, y)
```

```python
# Predictions
y_pred = model.predict(X_poly)
```

```python
# Evaluation
print("R2 Score: ", r2_score(y, y_pred))
print("RMSE: ", np.sqrt(mean_squared_error(y, y_pred)))
```

```python
# Plot
plt.scatter(x, y)
plt.plot(x, y_pred)
plt.title("Polynomial Regression (Degree 2)")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```
