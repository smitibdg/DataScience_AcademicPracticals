```python
import pandas as pd
import numpy as np
from scipy.stats import chisquare
```

```python
# Observed frequency
O_smiti = np.array([8,10 , 9 ,12 , 11 , 10]) # Observed from experiement
# Expected frequencies for a fair die
E_smiti = np.array([60/6] *6)    # 10  , 10 , 10 , 10 , 10 , 10
print("Observed :" , O_smiti)
print("Expected :" , E_smiti)

# Manual chi-square
chi_manual = np.sum((O_smiti-E_smiti)**2 / E_smiti)
print(f"Manual chi-square value : {chi_manual}")

# Using SCipy
chi_c , p_value = chisquare(f_obs=O_smiti , f_exp=E_smiti)
print(f"Scipy chi-square : {chi_c}")
print(f"P value : {p_value}")

alpha = 0.05
if p_value < alpha:
      print("\nConclusion Reject H0 (Null Hypo) (Observed frequencies do Not fit expected)")
else:
    print("\nconclusion Fail to Reject H0 (Observed frequencies fit expected distribution)")
```

```python
import numpy as np
from scipy.stats import kstest , norm

np.random.seed(1)

# Generate sample data (student's marsk)
data_smiti = np.random.normal(loc = 50 , scale = 10 ,size=100)

# Perform KS Test against N(50 , 10)
ks_stat, p_value = kstest(data_smiti, norm.cdf, args=(50, 10))

print("KS Statistcis :" , ks_stat)
print("p-value:" , p_value)

alpha = 0.05
if p_value < alpha:
    print("\n Conclusion : Reject H0 (Data does not follow the normal distribtution)")
else:
    print("\n Conclusion : Fail to reject H0 ( Data follow the normal distribution)")
```

```python
import numpy as np
from scipy.stats import chi2_contingency

# contiengency Table
# Rows = Gender , Columns  = Preferences

table_smiti = np.array([
    [30 ,20] ,
    [50 , 40]
])
print(f"Contigency Table :\n {table_smiti}")

# Chi square of Independences
chi2 , p_value , dof , expected = chi2_contingency(table_smiti)

print(f"\nchi-square Statistics :" , chi2)
print(f"Degree of freedom(DOF) :" , dof)
print(f"Expected Table \n :" , expected)
print(f"p-value \n :" , p_value)

alpha = 0.05
if p_value < alpha:
    print("\n Conclusion : Reject H0 ( Gender and Preferencex are dependent)")
else:
    print("\n Conclusion : Fail to Reject H0 ( No relationship Independent)")
```
