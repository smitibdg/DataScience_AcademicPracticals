Spatial series analysis studies whether values observed at nearby location are related.
If nearby regions hav similar values, we say there is spatial correction.

Hypothesis

HO: No spatial autocorrelation

H1: Spatial autocorrelation exists

```python
import pandas as pd
import numpy as np
import geopandas as gpd
from  shapely.geometry import Point
from libpysal.weights import Queen
from esda.moran import Moran
from spreg import ML_Lag
```

```python
np.random.seed(42)

n_smiti = 25
x_coords_smiti =  np.repeat(range(5),5)
y_coords_smiti = np.tile(range(5),5)
data_smiti = pd.DataFrame({
    'value' : np.random.normal(100,10,n_smiti),
    'income' : np.random.normal(50000,8000,n_smiti),
    'geometry' : [Point(xy) for xy in zip(x_coords_smiti,y_coords_smiti)]

})

gdf  = gpd.GeoDataFrame(data_smiti, geometry='geometry')
```

```python
w_smiti = Queen.from_dataframe(gdf)
w_smiti.transform = 'r'
```

```python
y_smiti = gdf['value'].values

moran_smiti = Moran(y_smiti, w_smiti)
print('morans i', moran_smiti.I)
print("Morans p_value", moran_smiti.sim)
```

p-value < 0.05 --> Reject Ho (spatial autocorrealtion exists)

p-value >= 0.05 --> No spatial autocorrelation

Hypothesis

Ho : p = 0 {no spatial dependecy)
H1 : p NOT 0 Spatial dependence

```python
y_smiti = gdf[['value']].values
x_smiti = gdf[['income']].values
ml = ML_Lag(y_smiti, x_smiti, w = w_smiti, name_y='value', name_x=['income'])
print(ml.summary)
```

```python
ml1_smiti = ML_Lag(y_smiti, x_smiti, w_smiti, name_y='value')
print(ml1_smiti.summary)
```

Decision rule :

If p < 0.05 Reject Ho

Computed p value = 0.0005

Which is < 0.05

Hence Reject Ho
