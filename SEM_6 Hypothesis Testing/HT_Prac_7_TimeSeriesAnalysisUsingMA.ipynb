```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create sample time series data
data_smiti = {
    'Month': pd.date_range(start='2023-01-01', periods=24, freq='M'),
    'Value': [120, 125, 130, 128, 135, 140, 145, 150, 148, 155, 360, 165,
              170, 175, 100, 178, 185, 190, 195, 200, 198, 205, 210, 215]
}

df_smiti = pd.DataFrame(data_smiti)
df_smiti.set_index('Month', inplace=True)
df_smiti
```

```python
plt.figure(figsize=(8, 4))
plt.plot(df_smiti['Value'], label='Original Data', marker='o')
plt.title('Time Series Data')
plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
df_smiti['MA_3'] = df_smiti['Value'].rolling(window=3).mean()
```

```python
df_smiti['MA_6'] = df_smiti['Value'].rolling(window=6).mean()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(df_smiti['Value'], label='Original Data', alpha=0.6)
plt.plot(df_smiti['MA_3'], label='3-Period Moving Average', color='red')
plt.plot(df_smiti['MA_6'], label='6-Period Moving Average', color='green')
plt.title('Moving Average Time Series Analysis')
plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
# Forecast next value using last 3 values
forecast_smiti = df_smiti['Value'].tail(3).mean()
print("Forecasted next value (3-paired MA): ", forecast_smiti)
```
