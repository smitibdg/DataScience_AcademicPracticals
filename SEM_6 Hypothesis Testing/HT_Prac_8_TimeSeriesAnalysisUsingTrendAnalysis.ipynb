```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sample yearly Sales Data
data_smiti = {
    'Year': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022],
    'Sales': [120, 135, 150, 165, 180, 200, 220, 250]
}

df_smiti = pd.DataFrame(data_smiti)
df_smiti
```

```python
plt.plot(df_smiti['Year'], df_smiti['Sales'])
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Original Time Series Data')
plt.show()
```

```python
# Calculate 3-Year Moving Average
df_smiti['Moving_Average'] = df_smiti['Sales'].rolling(window=3).mean()
df_smiti
```

```python
plt.plot(df_smiti['Year'], df_smiti['Sales'], label='Original Data')
plt.plot(df_smiti['Year'], df_smiti['Moving_Average'], label='Moving Average Trend')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Time Series Data with Moving Average Trend')
plt.legend()
plt.show()
```

```python
df_smiti['Time'] = np.arange(1, len(df_smiti) + 1)
```

```python
coefficients_smiti = np.polyfit(df_smiti['Time'], df_smiti['Sales'], 1)
trend_line_smiti = np.polyval(coefficients_smiti, df_smiti['Time'])

df_smiti['Trend_Line'] = trend_line_smiti
df_smiti
```

```python
plt.plot(df_smiti['Year'], df_smiti['Sales'], label='Original Data')
plt.plot(df_smiti['Year'], df_smiti['Trend_Line'], label='Linear Trend Line')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Time Analysis using Least Squares Method')
plt.legend()
plt.show()
```

```python
print("Trend Equation: ")
print(f"Y = {coefficients_smiti[0]:.2f}X + {coefficients_smiti[0]:.2f}")
```
