```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random. seed(0)

time_smiti = np.arange(0, 100)
signal_smiti = (
    10 * np.sin(2 * np.pi * time_smiti / 12) +   # yearly/seasonal cycle
    5 * np.sin(2 * np.pi * time_smiti / 6) +     # shorter cycle
    np.random.normal(0, 2, 100)            # noise
)

df_smiti = pd.DataFrame({'Time': time_smiti, 'Value': signal_smiti})
df_smiti
```

```python
plt.plot(df_smiti['Time'], df_smiti['Value'])
plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Time Series Data")
plt.show()
```

```python
fft_values_smiti = np.fft.fft(df_smiti['Value'])
frequencies_smiti = np.fft.fftfreq(len(fft_values_smiti))
```

```python
power_smiti = np.abs(fft_values_smiti) ** 2
```

```python
plt.plot(frequencies_smiti[:50], power_smiti[:50])
plt.xlabel("Frequency")
plt.ylabel("Power")
plt.title("Power Spectrum")
plt.show()
```
