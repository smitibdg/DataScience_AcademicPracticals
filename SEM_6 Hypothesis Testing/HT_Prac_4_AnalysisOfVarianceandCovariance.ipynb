```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.multivariate.manova import MANOVA
```

ONE WAY MANOVA

Testing whether 3 teaching methods produce different scores.

H0: All teachging methods have same mean

H1: Atleast one method differs

```python
np.random.seed(1)

data_smiti = pd.DataFrame({
    'Score': np.concatenate([
        np.random.normal(75, 5, 20),
        np.random.normal(80, 5, 20),
        np.random.normal(85, 5, 20)
    ]),
    'Method':['A']*20 + ['B']*20 + ['C']*20
})
```

```python
model_smiti = ols('Score ~ Method', data=data_smiti).fit()
anova_table_smiti = sm.stats.anova_lm(model_smiti, typ=2)
print(anova_table_smiti)
```

TWO WAY ANOVA

Testing Effect of:


*   Teaching Method
*   Gender
*   Interaction Effect

H0: No effect of Method, Gender, or Interaction

H1: Atleast one factor affects scores.







```python
np.random.seed(2)

data2_smiti = pd.DataFrame({
    'Score': np.random.normal(80, 5, 60),
    'Method':['A', 'B', 'C'] * 20,
    'Gender':['Male', 'Female'] * 30
})
```

```python
model2_smiti = ols('Score ~ Method + Gender + Method:Gender', data=data2_smiti).fit()
anova_table2_smiti = sm.stats.anova_lm(model2_smiti, typ=2)
print(anova_table_smiti)
```

MANOVA (Multivariate ANOVA)

Testing whether teaching methods affect:
*   Maths Score
*   Science Score
*   English Score


(Multiple dependent variables)
H0: Teaching methods does not affect combined subject scores
H1: Teaching method affects atleast one subject.

```python
np.random.seed(3)

manova_data_smiti = pd.DataFrame({
    'Maths': np.random.normal(75, 5, 60),
    'Science': np.random.normal(70, 5, 60),
    'English': np.random.normal(80, 5, 60),
    'Method': ['A', 'B', 'C'] * 20
})

manova_data_smiti
```

```python
manova_smiti = MANOVA.from_formula('Maths + Science + English ~ Method', data=manova_data_smiti)
print(manova_smiti.mv_test())
```

ANCOVA (Analysis of Covariance)


Testing teaching method effect while controlling study hours

H0: Method has no effect after controlling study hours

H1: Method affects scores even after controlling study hours

```python
np.random.seed(4)

ancova_data_smiti = pd.DataFrame({
    'Score': np.random.normal(80, 5, 60),
    'Method': ['A', 'B', 'C'] * 20,
    'StudyHours': np.random.normal(5, 1, 60)
})
```

```python
model3_smiti = ols('Score ~ Method + StudyHours', data=ancova_data_smiti).fit()
ancova_table3_smiti = sm.stats.anova_lm(model3_smiti, typ=2)

print(ancova_table3_smiti)
```
