```python
!pip install statsmodels
import numpy as np
import pandas as pd
from scipy import stats
import statsmodels.stats.weightstats as ztest
```

```python
# First Generate sample data
np.random.seed(1) # 1 indicate nothing
sample = np.random.normal(loc=168, scale=5, size=30) # mean 168 and S.D. is 5

# Population mean to test
mu_0 = 165

# Perform one-sample t-test
t_stat, p_val = stats.ttest_1samp(sample, mu_0)

print("One Sample T-test")
print("T-test statistics: ", t_stat)
print("p_value: ", p_val)

alpha = 0.05
if p_val < alpha:
  print("Reject the H0 --> Mean is significantly different from 165")
else:
  print("Do NOT Reject H0 --> No significant different from 165")
```

```python
# Generate two Tndependent Samples
np.random.seed(10)
boys = np.random.normal(loc=70, scale=8, size=35)
girls = np.random.normal(loc=75, scale=7, size=30)

# Perform Two-Sample T-test (Independent)
t_stat, p_val = stats.ttest_ind(boys, girls)

print("Two-Sample T-test")
print("T-test statistics: ", t_stat)
print("p_value: ", p_val)

alpha = 0.05
if p_val < alpha:
  print("Reject the H0 --> Boys and Girls Mean marks are different")
else:
  print("Do NOT Reject H0 --> No difference in Means")
```

```python
# Generate paired data (before & after)
np.random.seed(20)
before = np.random.normal(loc=70, scale=5, size=25)
after = before - np.random.normal(loc=2, scale=1, size=25)

# Paired T-test
t_stat, p_val = stats.ttest_rel(before, after)

print("Paired T-test")
print("T-test statistics: ", t_stat)
print("p_value: ", p_val)

alpha = 0.05
if p_val < alpha:
  print("Reject the H0 --> Significant before-after difference")
else:
  print("Do NOT Reject H0 --> No Significant change")
```

```python
# Generate Rate
np.random.seed(100)
scores = np.random.normal(loc=60, scale=10, size=50) #mean 60 and S.D. 10

# Perform Z-test
z_stat, p_val = ztest.ztest(scores, value=60)

print("Z-test")
print("Z-test statistics: ", z_stat)
print("p_value: ", p_val)

alpha = 0.05
if p_val < alpha:
  print("Reject the H0 --> Mean is significantly different from 60")
else:
  print("Do NOT Reject H0 --> No significant different from 60")
```
